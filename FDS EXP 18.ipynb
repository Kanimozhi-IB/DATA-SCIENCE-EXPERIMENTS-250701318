```python

import matplotlib.pyplot as plt

# Sample data
students = ['A', 'B', 'C', 'D', 'E']
marks = [85, 70, 90, 65, 80]
attendance = [90, 75, 95, 60, 85]

# Create multiple visualizations
plt.figure(figsize=(12, 8))

# 1. Line Plot
plt.subplot(2, 2, 1)
plt.plot(students, marks, marker='o')
plt.title("Line Plot - Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

# 2. Bar Chart
plt.subplot(2, 2, 2)
plt.bar(students, marks)
plt.title("Bar Chart - Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

# 3. Scatter Plot
plt.subplot(2, 2, 3)
plt.scatter(attendance, marks)
plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

# 4. Pie Chart
plt.subplot(2, 2, 4)
plt.pie(marks, labels=students, autopct='%1.1f%%')
plt.title("Marks Distribution")

plt.tight_layout()
plt.show()
```
