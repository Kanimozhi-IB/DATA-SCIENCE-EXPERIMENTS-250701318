```python

import matplotlib.pyplot as plt

# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
marks = [45, 50, 60, 55, 70, 85]

# Create line plot
plt.plot(months, marks, marker='o', label='Student Marks')

# Add title and labels
plt.title('Monthly Student Performance')
plt.xlabel('Months')
plt.ylabel('Marks')

# Display grid and legend
plt.grid(True)
plt.legend()

# Display graph
plt.show()
```

```python

import matplotlib.pyplot as plt

# Sample data
subjects = ['Maths', 'Physics', 'Chemistry', 'English', 'Computer']
marks = [75, 65, 80, 85, 90]

# Create bar chart
plt.bar(subjects, marks, label='Student Marks')

# Add title and labels
plt.title('Student Marks in Different Subjects')
plt.xlabel('Subjects')
plt.ylabel('Marks')

# Display grid and legend
plt.grid(axis='y')
plt.legend()

# Display chart
plt.show()
```

```python

import matplotlib.pyplot as plt

# Sample data
activities = ['Study', 'Sleep', 'Exercise', 'Entertainment', 'Others']
hours = [6, 8, 2, 4, 4]

# Create pie chart
plt.pie(hours, labels=activities, autopct='%1.1f%%',
        startangle=90)

# Add title
plt.title('Daily Time Distribution')

# Display chart
plt.show()
```
